```python
#Install Packages
import pandas as pd

#Load in data
air_data = pd.read_csv('AirQuality_Daily_StudentVersion.csv')

#Put data in a dataframe
air_data = pd.DataFrame(air_data)

#Check that data is properly loaded
print(air_data)
```

```python
###Task 1

##Group by location and calculate the mean and median concentrations of pollutants
stat_summary = air_data.groupby(['sensor.name']).agg(pm25_mean = ('pm2.5_atm', 'mean'),
                                                     pm25_median = ('pm2.5_atm', 'median'),
                                                     pm10_mean = ('pm10.0_atm', 'mean'),
                                                     pm10_median = ('pm10.0_atm', 'median'),
                                                     voc_mean = ('voc', 'mean'),
                                                     voc_median = ('voc', 'median')).round(2)

##Put in ascending order and display the top five values of each

#PM 2.5
top5_pm25_mean = stat_summary.sort_values(by='pm25_mean', ascending=False).head(5)
top5_pm25_median = stat_summary.sort_values(by='pm25_median', ascending=False).head(5)

#PM 10.0
top5_pm10_mean = stat_summary.sort_values(by='pm10_mean', ascending=False).head(5)
top5_pm10_median = stat_summary.sort_values(by='pm10_median', ascending=False).head(5)

#VOC
top5_voc_mean = stat_summary.sort_values(by='voc_mean', ascending=False).head(5)
top5_voc_median = stat_summary.sort_values(by='voc_median', ascending=False).head(5)

#Display the information
print("Top 5 PM 2.5 Locations by Mean")
display(top5_pm25_mean)

print("Top 5 PM 2.5 Locations by Median")
display(top5_pm25_median)

print("Top 5 PM 10.0 Locations by Mean")
display(top5_pm10_mean)

print("Top 5 PM 10.0 Locations by Median")
display(top5_pm10_median)

print("Top 5 VOC Locations by Mean")
display(top5_voc_mean)

print("Top 5 VOC Locations by Median")
display(top5_voc_median)
```

```python
###Task 2

#Get the max values of each pollutant
pm25_max_value = air_data.agg(pm25_max=('pm2.5_atm','max')).iloc[0,0]
pm10_max_value = air_data.agg(pm10_max=('pm10.0_atm','max')).iloc[0,0]
voc_max_value = air_data.agg(voc_max=('voc','max')).iloc[0,0]

#Filter to the rows where pollutant levels are at maximums
pm25_max_rows = air_data[air_data['pm2.5_atm'] == pm25_max_value]
pm10_max_rows = air_data[air_data['pm10.0_atm'] == pm10_max_value]
voc_max_rows = air_data[air_data['voc'] == voc_max_value]

#Display the PM 2.5 values
pm25_max_rows
```

```python
###Task 3

##Categorize temperatures
#Step 1 - Define a function that assigns temperature levels
def assign_temperature(temperature):
    if temperature < 32:
        return 'Below Freezing'
    elif 32 <= temperature <= 50:
        return 'Cool'
    elif 50 <= temperature <= 70:
        return 'Warm'
    else:
        return 'Hot'

#Step 2 - Apply the function to my data
air_data['temperature_level'] = air_data['temperature'].apply(assign_temperature)  
#the input to my assign_temperature function would be the data column temperature

air_data

##Categorize humidities
#Step 1 - Define a function that assigns humidity levels
def assign_humidity(humidity):
    if humidity < 50:
        return 'Low'
    elif 50 <= humidity <= 80:
        return 'High'
    else:
        return 'Very High'

#Step 2 - Apply the function to my data
air_data['humidity_level'] = air_data['humidity'].apply(assign_humidity)  
#the input to my assign_temperature function would be the data column temperature

air_data

##make two pivot tables with means, one will be hastagged because two can't display at once
temperature_stats = pd.pivot_table(air_data,index="temperature_level",values=["pm2.5_atm","pm10.0_atm","voc"],aggfunc=["mean"])

temperature_stats
```

```python
###Task 3 Continued

#added a new code block here for the sole purpose of showing the humidity pivot table

humidity_stats = pd.pivot_table(air_data,index="humidity_level",values=["pm2.5_atm","pm10.0_atm","voc"],aggfunc=["mean"])

humidity_stats
```

```python
###Task 4
##PM 2.5
#Step 1 - Isolate unhealthy levels of PM 2.5 
unhealthy_pm25 = air_data[(air_data['pm2.5_atm']>=35.5) &
                          (air_data['pm2.5_atm']<=55.4)]

#Step 2 - sort by date, sensor name, and location
unhealthy_pm25_events = unhealthy_pm25[['date','sensor.name','pm2.5_atm','sensor.latitude','sensor.longitude']].sort_values('date')  

unhealthy_pm25_events.head(3)
```

```python
###Task 4 Continued

##PM 10.0
#Step 1 - Isolate unhealthy levels of PM 10.0, anything greater than 100
unhealthy_pm10 = air_data[(air_data['pm10.0_atm']>=155) &
                          (air_data['pm10.0_atm']<=254)]

#Step 2 - sort by date and sensor name
unhealthy_pm10_events = unhealthy_pm10[['date','sensor.name','pm10.0_atm','sensor.latitude','sensor.longitude']].sort_values('date')  

unhealthy_pm10_events
```
